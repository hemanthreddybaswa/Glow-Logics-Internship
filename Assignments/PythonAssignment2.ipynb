
Mini Project 1: Amazon User Profile

```python
user_profile = {
    "name": "Hemanth",
    "email": "hemanth@gmail.com",
    "orders": ["Book - Python Programming", "Headphones", "Coffee Mug"],
    "prime_member": True,
    "city": "Vizag"
}

print("----------------------------------------")
print(f"Welcome back, {user_profile['name']}!")
print("Your Amazon Dashboard")
print("----------------------------------------")
print(f"Email: {user_profile['email']}")
print(f"City: {user_profile['city']}")
print(f"Prime Membership: {'Active' if user_profile['prime_member'] else 'Inactive'}")
print("Recent Orders:")
if user_profile['orders']:
    for order in user_profile['orders']:
        print(f"  - {order}")
else:
    print("  No recent orders.")
print("----------------------------------------")
```

Mini Project 2: Student Management System

```python
print("\n--- Student Information Collection ---")
student_name = input("Enter student's name: ")
student_id = input("Enter student ID: ")
student_major = input("Enter student's major: ")
student_gpa = float(input("Enter student's GPA: "))

student_profile = {
    "Name": student_name,
    "ID": student_id,
    "Major": student_major,
    "GPA": student_gpa
}

print("----------------------------------------")
for key, value in student_profile.items():
    print(f"{key:}: {value}")
print("----------------------------------------")
```
